```python
class Baseball_Team:
    def __init__(self, name, win, lose, draw):
        self.name = name
        self.win = win
        self.lose = lose
        self.draw = draw

    # 勝率を計算するメソッド
    def calc_win_rate(self):
        if self.win + self.lose == 0:
            return 0
        return self.win / (self.win + self.lose)

    # 勝敗情報を表示するメソッド
    def show_team_result(self):
        win_rate = self.calc_win_rate()
        print(f"{self.name:10s} {self.win:3d} {self.lose:3d} {self.draw:3d} {win_rate:.3f}")

# 6チームのインスタンス生成
team_list = [
    Baseball_Team("Archers", 77, 64, 2),
    Baseball_Team("Meteors", 80, 61, 2),
    Baseball_Team("Cheetahs", 74, 68, 1),
    Baseball_Team("Sharks", 69, 73, 1),
    Baseball_Team("Pythons", 65, 76, 2),
    Baseball_Team("Owls",    60, 81, 2)
    ]

# 見出しの表示
print("Team       Win Lose Draw WinRate")

# 各チームの結果表示
for team in team_list:
    team.show_team_result()

```
