```python
# 空のリストを作成
words = []

# アルファベット小文字の各文字をキー、値を0とする辞書を作成
counts = {chr(c): 0 for c in range(ord('a'), ord('z') + 1)}

# 英単語の入力受付
while True:
    word = input("英単語を入力してください：").strip()
    if word == "":
        break

    # 入力された英単語をリストに追加
    words.append(word)

    # 各文字をカウント
    for ch in word.lower():
        if ch in counts:
            counts[ch] += 1

# --- 結果表示 ---

# 入力された英単語を ABC 順に並べ替えて表示
print(f"入力した英単語：{sorted(words)}")

# 1回以上登場した文字だけを表示
for ch, num in counts.items():
    if num > 0:
        print(f"{ch}が{num}個ありました")

```
